```python
%pip install pandas matplotlib

import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("data.csv", sep=";")

data.head()
```

```python
print("Number of students:", data.shape[0])
print("Number of variables:", data.shape[1])
```

```python
print("Columns in the dataset:")

for column in data.columns:
    print("-", column)
```

```python
print("Missing values:")
print(data.isnull().sum())
```

```python
duplicates = data.duplicated().sum()

print("Number of duplicate records:", duplicates)
```

```python
print("Student Outcome Distribution:")
print(data["Target"].value_counts())
```

```python
data.describe()
```

```python
import matplotlib.pyplot as plt

outcome_counts = data["Target"].value_counts()

plt.figure(figsize=(8, 5))
outcome_counts.plot(kind="bar")

plt.title("Student Outcome Distribution")
plt.xlabel("Student Outcome")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(data["Age at enrollment"], bins=15)

plt.title("Age Distribution of Students")
plt.xlabel("Age at Enrollment")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(data["Admission grade"], bins=15)

plt.title("Distribution of Admission Grades")
plt.xlabel("Admission Grade")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

data.boxplot(
    column="Curricular units 2nd sem (grade)",
    by="Target"
)

plt.title("Academic Performance by Student Outcome")
plt.suptitle("")
plt.xlabel("Student Outcome")
plt.ylabel("Second Semester Grade")

plt.tight_layout()
plt.show()
```

```python
approved = data.groupby("Target")[
    "Curricular units 2nd sem (approved)"
].mean()

plt.figure(figsize=(8, 5))

approved.plot(kind="bar")

plt.title("Average Courses Approved by Student Outcome")
plt.xlabel("Student Outcome")
plt.ylabel("Average Number of Approved Courses")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
scholarship = data["Scholarship holder"].value_counts()

plt.figure(figsize=(7, 5))

scholarship.plot(kind="bar")

plt.title("Scholarship Status of Students")
plt.xlabel("Scholarship Status")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
tuition = data["Tuition fees up to date"].value_counts()

plt.figure(figsize=(7, 5))

tuition.plot(kind="bar")

plt.title("Students with Tuition Fees Up to Date")
plt.xlabel("Tuition Fee Status")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
semester_performance = pd.DataFrame({
    "First Semester": [
        data["Curricular units 1st sem (grade)"].mean()
    ],
    "Second Semester": [
        data["Curricular units 2nd sem (grade)"].mean()
    ]
})

plt.figure(figsize=(7, 5))

semester_performance.iloc[0].plot(kind="bar")

plt.title("Average Student Performance by Semester")
plt.xlabel("Semester")
plt.ylabel("Average Grade")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
